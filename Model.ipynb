```python
# ============================================================
# MODULE 2: DATASET AUTO-DISCOVERY & ARCHIVE EXTRACTION
# ============================================================

# Search for existing Dataset directory
CANDIDATE_DIRS = [
    os.path.join(os.getcwd(), "Dataset"),
    r"c:\AgroGuard\Dataset",
    os.path.join(os.path.dirname(os.getcwd()), "Dataset"),
    r"c:\AgroGuard\Kaggle\Dataset"
]

DATASET_DIR = None
for cand in CANDIDATE_DIRS:
    if os.path.exists(cand) and os.path.exists(os.path.join(cand, "train")):
        DATASET_DIR = cand
        print(f"✅ Found active Dataset directory: {DATASET_DIR}")
        break

# If not found, search for Dataset.zip and auto-unzip
if not DATASET_DIR:
    CANDIDATE_ZIPS = [
        os.path.join(os.getcwd(), "Dataset.zip"),
        r"c:\AgroGuard\Dataset.zip",
        r"c:\AgroGuard\Kaggle\Dataset.zip",
        os.path.join(os.path.dirname(os.getcwd()), "Dataset.zip")
    ]
    
    zip_to_extract = None
    for z in CANDIDATE_ZIPS:
        if os.path.exists(z):
            zip_to_extract = z
            print(f"📦 Found archive: {z}")
            break
            
    if zip_to_extract:
        target_extract = os.path.join(os.getcwd(), "Dataset")
        print(f"⏳ Extracting {os.path.basename(zip_to_extract)} to {target_extract}...")
        with zipfile.ZipFile(zip_to_extract, 'r') as zf:
            zf.extractall(os.getcwd())
        if os.path.exists(target_extract) and os.path.exists(os.path.join(target_extract, "train")):
            DATASET_DIR = target_extract
            print(f"✅ Successfully extracted dataset to: {DATASET_DIR}")
    else:
        raise FileNotFoundError("Could not find Dataset folder or Dataset.zip. Please verify dataset location.")

# Verify splits
train_dir = os.path.join(DATASET_DIR, "train")
val_dir   = os.path.join(DATASET_DIR, "val")
test_dir  = os.path.join(DATASET_DIR, "test")

train_classes = sorted([d for d in os.listdir(train_dir) if os.path.isdir(os.path.join(train_dir, d))])
val_classes   = sorted([d for d in os.listdir(val_dir) if os.path.isdir(os.path.join(val_dir, d))])
test_classes  = sorted([d for d in os.listdir(test_dir) if os.path.isdir(os.path.join(test_dir, d))])

print(f"📊 Dataset Splits Verified:")
print(f"   • Train Classes: {len(train_classes)} ({sum(len(os.listdir(os.path.join(train_dir, c))) for c in train_classes)} images)")
print(f"   • Val Classes  : {len(val_classes)} ({sum(len(os.listdir(os.path.join(val_dir, c))) for c in val_classes)} images)")
print(f"   • Test Classes : {len(test_classes)} ({sum(len(os.listdir(os.path.join(test_dir, c))) for c in test_classes)} images)")
assert train_classes == val_classes == test_classes, "Class lists between splits must match identically!"
```

```python
# ============================================================
# MODULE 3: HIGH-PERFORMANCE HYPERPARAMETER ENGINE
# ============================================================

# 1. Adaptive Batch Size based on GPU VRAM
if torch.cuda.is_available():
    if vram_gb >= 11.0:
        BATCH_SIZE = 64
    elif vram_gb >= 6.0:
        BATCH_SIZE = 32
    else:
        BATCH_SIZE = 16
else:
    BATCH_SIZE = 16

# On Windows notebooks, num_workers=0 avoids multiprocessing spawn crashes
NUM_WORKERS = 0 if os.name == 'nt' else min(4, os.cpu_count() or 2)
PIN_MEMORY = torch.cuda.is_available()
USE_AMP = torch.cuda.is_available()  # Automatic Mixed Precision for 2x faster GPU training

CONFIG = {
    'architecture': 'ConvNeXt-Tiny',
    'backbone': 'convnext_tiny.fb_in22k_ft_in1k',
    'img_size': 256,
    'batch_size': BATCH_SIZE,
    'num_workers': NUM_WORKERS,
    'pin_memory': PIN_MEMORY,
    'use_amp': USE_AMP,
    'lr': 1e-4,
    'min_lr': 1e-6,
    'weight_decay': 1e-2,
    'epochs': 10,
    'label_smoothing': 0.1,
    'output_model': 'best_agroguard_model.pth'
}

# Empirical Agricultural leaf normalization
AGRO_MEAN = [0.4678, 0.4931, 0.3542]
AGRO_STD  = [0.2114, 0.1983, 0.2248]

print("⚙️ Active Training Configuration:")
for k, v in CONFIG.items():
    print(f"   • {k:<16}: {v}")
```

  <div class='agro-badge'>🌿 Field Robustness & Class Balancing</div>
  <div class='agro-title'>Module 4: Sunlight, High-Contrast & Shadow Augmentation + Balanced Sampling</div>
  <div class='agro-subtitle'>Engineered for field deployment under harsh tropical sunlight, glare, deep canopy shade, and 550:1 class imbalance using WeightedRandomSampler and smoothed loss weights.</div>
</div>


```python
# ============================================================
# MODULE 4: FIELD-ROBUST AUGMENTATION & BALANCED DATALOADERS
# Robust to Harsh Sunlight, Specular Glare, High Contrast, & Canopy Shadows
# Balanced across 48 Classes with Weighted Sampling & Effective-Loss Weights
# ============================================================
from torchvision import datasets, transforms
from torch.utils.data import DataLoader, WeightedRandomSampler
from collections import Counter
import numpy as np

# ── Field-Resilient Sunlight, Glare, Shadow, & Contrast Augmentations ──
# Mimics open-sky tropical field conditions: harsh midday sun, canopy shadows,
# overexposed specular glare on waxy leaves, and handheld phone angle shifts.
train_transform = transforms.Compose([
    transforms.RandomResizedCrop(CONFIG['img_size'], scale=(0.70, 1.0), interpolation=transforms.InterpolationMode.BILINEAR),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.5),
    transforms.RandomAffine(degrees=20, translate=(0.08, 0.08), scale=(0.90, 1.10), shear=8),
    # Harsh sunlight vs deep shade: dynamic contrast & brightness expansion
    transforms.ColorJitter(brightness=0.35, contrast=0.35, saturation=0.30, hue=0.08),
    # Overcome ambient illumination shifts (direct sun vs diffuse clouds)
    transforms.RandomAutocontrast(p=0.4),
    transforms.RandomEqualize(p=0.2),
    # Specular sun reflection / cuticle solarization simulation
    transforms.RandomSolarize(threshold=200, p=0.15),
    # Camera lens dust / optical glare resilience
    transforms.RandomApply([transforms.GaussianBlur(kernel_size=(3, 3), sigma=(0.1, 1.5))], p=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=AGRO_MEAN, std=AGRO_STD),
    # Dappled canopy shadows & partial leaf occlusion simulation
    transforms.RandomErasing(p=0.25, scale=(0.02, 0.20), ratio=(0.4, 2.5), value='random')
])

eval_transform = transforms.Compose([
    transforms.Resize((CONFIG['img_size'], CONFIG['img_size']), interpolation=transforms.InterpolationMode.BILINEAR),
    transforms.ToTensor(),
    transforms.Normalize(mean=AGRO_MEAN, std=AGRO_STD)
])

train_dataset = datasets.ImageFolder(train_dir, transform=train_transform)
val_dataset   = datasets.ImageFolder(val_dir, transform=eval_transform)
test_dataset  = datasets.ImageFolder(test_dir, transform=eval_transform)

CLASS_NAMES = train_dataset.classes
NUM_CLASSES = len(CLASS_NAMES)

# ── Class Imbalance Mitigation: Dual-Stage Balancing ──
# 1. WeightedRandomSampler: Ensures minority classes (e.g., Orange Scab with 8 images)
#    are sampled frequently in every batch alongside dominant classes.
# 2. Balanced Loss Weights: Dynamically weights cross-entropy gradient penalization.
targets = train_dataset.targets
class_counts = Counter(targets)
counts_per_class = [class_counts[i] for i in range(NUM_CLASSES)]

min_cnt, max_cnt = min(counts_per_class), max(counts_per_class)
print(f" Training Set Class Distribution:")
print(f"    Min Class Samples : {min_cnt} ({CLASS_NAMES[int(np.argmin(counts_per_class))]})")
print(f"    Max Class Samples : {max_cnt} ({CLASS_NAMES[int(np.argmax(counts_per_class))]})")
print(f"    Imbalance Ratio   : {max_cnt / max(1, min_cnt):.1f}:1")

# Reciprocal frequency sample weights for DataLoader sampler (power-smoothed at 0.6)
class_sample_weights = [1.0 / (max(1, count) ** 0.6) for count in counts_per_class]
sample_weights = torch.tensor([class_sample_weights[t] for t in targets], dtype=torch.float32)

train_sampler = WeightedRandomSampler(
    weights=sample_weights,
    num_samples=len(train_dataset),
    replacement=True
)

# Smoothed class loss weights for CrossEntropyLoss
raw_weights = [(max_cnt / max(1, c)) ** 0.4 for c in counts_per_class]
class_weights = torch.tensor(raw_weights, dtype=torch.float32)
class_weights = (class_weights / class_weights.mean()).to(device)

print(f" Class Balancing Active:")
print(f"    WeightedRandomSampler: ACTIVE (smooth=0.6, samples={len(train_dataset)})")
print(f"    Loss Weights Range   : [{class_weights.min().item():.3f} - {class_weights.max().item():.3f}] (mean={class_weights.mean().item():.3f})")

train_loader = DataLoader(
    train_dataset,
    batch_size=CONFIG['batch_size'],
    sampler=train_sampler,  # Balanced sampling eliminates minority class starvation
    num_workers=CONFIG['num_workers'],
    pin_memory=CONFIG['pin_memory'],
    drop_last=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=CONFIG['batch_size'],
    shuffle=False,
    num_workers=CONFIG['num_workers'],
    pin_memory=CONFIG['pin_memory']
)

test_loader = DataLoader(
    test_dataset,
    batch_size=CONFIG['batch_size'],
    shuffle=False,
    num_workers=CONFIG['num_workers'],
    pin_memory=CONFIG['pin_memory']
)

print(f" DataLoaders Initialized with Sunlight/Shadow Augmentations & Balanced Sampling!")
print(f"    Number of Classes : {NUM_CLASSES}")
print(f"    Train Batches     : {len(train_loader)} ({len(train_dataset)} samples)")
print(f"    Val Batches       : {len(val_loader)} ({len(val_dataset)} samples)")
print(f"    Test Batches      : {len(test_loader)} ({len(test_dataset)} samples)")
```

```python
# ============================================================
# MODULE 7: OPTIMIZATION ENGINE & TRAINING LOOP (LIVE TQDM)
# ============================================================
import torch.optim as optim
from tqdm.auto import tqdm

criterion = nn.CrossEntropyLoss(weight=class_weights, label_smoothing=CONFIG['label_smoothing'])
optimizer = optim.AdamW(model.parameters(), lr=CONFIG['lr'], weight_decay=CONFIG['weight_decay'])
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=CONFIG['epochs'], eta_min=CONFIG['min_lr'])
scaler = torch.cuda.amp.GradScaler(enabled=CONFIG['use_amp'])

history = {
    'train_loss': [], 'train_acc': [],
    'val_loss': [], 'val_acc': []
}

best_val_acc = 0.0
save_path = CONFIG['output_model']

print(f"🚀 Commencing ConvNeXt-Tiny Training ({CONFIG['epochs']} Epochs)...")
print("-" * 80)

for epoch in range(1, CONFIG['epochs'] + 1):
    start_time = time.time()
    
    # ── 1. Training Phase ──
    model.train()
    train_loss, train_correct, train_total = 0.0, 0, 0
    pbar = tqdm(train_loader, desc=f"Epoch [{epoch:02d}/{CONFIG['epochs']:02d}] Train", leave=False)
    for imgs, targets in pbar:
        imgs, targets = imgs.to(device, non_blocking=True), targets.to(device, non_blocking=True)
        optimizer.zero_grad()
        
        with torch.cuda.amp.autocast(enabled=CONFIG['use_amp']):
            outputs = model(imgs)
            loss = criterion(outputs, targets)
            
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        
        train_loss += loss.item() * imgs.size(0)
        _, preds = outputs.max(1)
        train_correct += preds.eq(targets).sum().item()
        train_total += imgs.size(0)
        
        pbar.set_postfix({
            'loss': f"{train_loss / train_total:.4f}",
            'acc': f"{100.0 * train_correct / train_total:.1f}%"
        })
        
    scheduler.step()
    epoch_train_loss = train_loss / train_total
    epoch_train_acc  = 100.0 * train_correct / train_total
    
    # ── 2. Validation Phase ──
    model.eval()
    val_loss, val_correct, val_total = 0.0, 0, 0
    val_pbar = tqdm(val_loader, desc=f"Epoch [{epoch:02d}/{CONFIG['epochs']:02d}] Val  ", leave=False)
    with torch.no_grad():
        for imgs, targets in val_pbar:
            imgs, targets = imgs.to(device, non_blocking=True), targets.to(device, non_blocking=True)
            with torch.cuda.amp.autocast(enabled=CONFIG['use_amp']):
                outputs = model(imgs)
                loss = criterion(outputs, targets)
                
            val_loss += loss.item() * imgs.size(0)
            _, preds = outputs.max(1)
            val_correct += preds.eq(targets).sum().item()
            val_total += imgs.size(0)
            
            val_pbar.set_postfix({
                'loss': f"{val_loss / val_total:.4f}",
                'acc': f"{100.0 * val_correct / val_total:.1f}%"
            })
            
    epoch_val_loss = val_loss / val_total
    epoch_val_acc  = 100.0 * val_correct / val_total
    elapsed = time.time() - start_time
    
    history['train_loss'].append(epoch_train_loss)
    history['train_acc'].append(epoch_train_acc)
    history['val_loss'].append(epoch_val_loss)
    history['val_acc'].append(epoch_val_acc)
    
    # Checkpoint logic
    saved_flag = ""
    if epoch_val_acc > best_val_acc:
        best_val_acc = epoch_val_acc
        torch.save(model.state_dict(), save_path)
        saved_flag = " ⭐ [BEST CHECKPOINT SAVED]"
        
    print(f"Epoch [{epoch:02d}/{CONFIG['epochs']:02d}] ({elapsed:.1f}s) | "
          f"Train Loss: {epoch_train_loss:.4f} Acc: {epoch_train_acc:.2f}% | "
          f"Val Loss: {epoch_val_loss:.4f} Acc: {epoch_val_acc:.2f}%{saved_flag}")

print("=" * 80)
print(f"🎉 Training Complete! Peak Validation Accuracy: {best_val_acc:.2f}%")
print(f"💾 Best Model Weights Serialized: {save_path}")

# Export unified_classes.json and dataset_stats.json for the AgroGuard Web Engine
with open('unified_classes.json', 'w', encoding='utf-8') as f:
    json.dump(CLASS_NAMES, f, indent=2)
stats_dict = {
    'total_classes': NUM_CLASSES,
    'classes': CLASS_NAMES,
    'total_images': len(train_dataset) + len(val_dataset) + len(test_dataset),
    'mean': AGRO_MEAN,
    'std': AGRO_STD
}
with open('dataset_stats.json', 'w', encoding='utf-8') as f:
    json.dump(stats_dict, f, indent=2)
print("📁 Exported 'unified_classes.json' and 'dataset_stats.json' for seamless web app synchronization.")

# Export unified_classes.json and dataset_stats.json for the AgroGuard Web Engine
with open('unified_classes.json', 'w', encoding='utf-8') as f:
    json.dump(CLASS_NAMES, f, indent=2)
stats_dict = {
    'total_classes': NUM_CLASSES,
    'classes': CLASS_NAMES,
    'total_images': len(train_dataset) + len(val_dataset) + len(test_dataset),
    'mean': AGRO_MEAN,
    'std': AGRO_STD
}
with open('dataset_stats.json', 'w', encoding='utf-8') as f:
    json.dump(stats_dict, f, indent=2)
print("📁 Exported 'unified_classes.json' and 'dataset_stats.json' for seamless web app synchronization.")
```

```python
# ============================================================
# MODULE 9: 3-VIEW TEST-TIME AUGMENTATION (TTA) EVALUATION
# ============================================================

def predict_tta(model, imgs):
    """Multi-view Test-Time Augmentation (Original, H-Flip, V-Flip)."""
    with torch.no_grad():
        with torch.cuda.amp.autocast(enabled=CONFIG['use_amp']):
            p1 = torch.softmax(model(imgs), dim=1)
            p2 = torch.softmax(model(torch.flip(imgs, dims=[3])), dim=1)
            p3 = torch.softmax(model(torch.flip(imgs, dims=[2])), dim=1)
        return (p1 + p2 + p3) / 3.0

# Load best weights
model.load_state_dict(torch.load(save_path, map_location=device))
model.eval()

all_preds = []
all_targets = []
top5_correct = 0
total_samples = 0

print("🔬 Running 3-View TTA Inference on Unseen Test Split (4,763 images)...")
with torch.no_grad():
    for imgs, targets in test_loader:
        imgs = imgs.to(device)
        probs = predict_tta(model, imgs).cpu()
        
        _, top1 = probs.max(1)
        _, top5 = probs.topk(5, dim=1)
        
        all_preds.extend(top1.numpy())
        all_targets.extend(targets.numpy())
        
        for t, k5 in zip(targets, top5):
            if t in k5:
                top5_correct += 1
        total_samples += len(targets)

top1_acc = 100.0 * np.sum(np.array(all_preds) == np.array(all_targets)) / total_samples
top5_acc = 100.0 * top5_correct / total_samples

print("=" * 80)
print(f"🏆 AGROGUARD UNIFIED MODEL TEST SET BENCHMARK:")
print(f"   • Total Test Samples  : {total_samples}")
print(f"   • Top-1 Accuracy      : {top1_acc:.2f}%")
print(f"   • Top-5 Accuracy      : {top5_acc:.2f}%")
print("=" * 80)
```
