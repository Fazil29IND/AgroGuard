```python
# ============================================================
# MODULE 1: COLAB CLOUD GPU & STORAGE DISCOVERY
# ============================================================
import os
import sys
import glob
import shutil
import zipfile
import time
import json
import warnings
warnings.filterwarnings('ignore')

import torch
import torch.nn as nn

print("=" * 80)
print("  AgroGuard AI v1.0: Google Colab Cloud GPU Unified Vision Pipeline")
print("=" * 80)

OUTPUT_DIR = '/content'

if torch.cuda.is_available():
    device = torch.device('cuda')
    gpu_name = torch.cuda.get_device_name(0)
    vram_gb = torch.cuda.get_device_properties(0).total_memory / (1024**3)
    torch.backends.cudnn.benchmark = True
    print(f"✅ Google Colab CUDA GPU Accelerator Detected!")
    print(f"   Device Name   : {gpu_name}")
    print(f"   Dedicated VRAM: {vram_gb:.2f} GB")
    print(f"   PyTorch Vers  : {torch.__version__}")
else:
    device = torch.device('cpu')
    vram_gb = 0.0
    print("⚠️ GPU not enabled. In Colab, go to: Runtime -> Change runtime type -> T4 GPU (Free).")

print(f"   Working Dir   : {OUTPUT_DIR}")
print("=" * 80)
```

<div class='agro-card'>
  <div class='agro-badge'>📂 Dataset Extraction</div>
  <div class='agro-title'>Module 2: Dataset Acquisition & Google Drive Mount</div>
  <div class='agro-subtitle'>Unzips Dataset.zip containing 51,320 images across 48 classes into /content/Dataset. Optional Google Drive mounting enables persistent cloud storage.</div>

```python
# ============================================================
# MODULE 2: DATASET ACQUISITION & EXTRACTION IN COLAB
# ============================================================

# Optional: Mount Google Drive if you keep Dataset.zip on your Drive
# Uncomment the two lines below to mount:
# from google.colab import drive
# drive.mount('/content/drive')

DATASET_DIR = '/content/Dataset'

def find_dataset_zip():
    # Check common locations in Colab
    candidates = [
        '/content/Dataset.zip',
        '/content/Colab/Dataset.zip',
        '/content/AgroGuard/Colab/Dataset.zip',
        '/content/AgroGuard/Dataset.zip',
        '/content/AgroGuard/Kaggle/Dataset.zip',
        os.path.join(os.getcwd(), 'Dataset.zip'),
        os.path.join(os.getcwd(), 'Colab/Dataset.zip')
    ]
    for p in candidates:
        if os.path.isfile(p):
            return p
            
    # Search /content and /content/drive recursively if needed
    for root, _, files in os.walk('/content'):
        for f in files:
            if f.lower().endswith('.zip') and ('dataset' in f.lower() or 'agro' in f.lower()):
                return os.path.join(root, f)
    return None

if not (os.path.exists(DATASET_DIR) and os.path.exists(os.path.join(DATASET_DIR, 'train'))):
    archive_path = find_dataset_zip()
    if archive_path:
        print(f"📦 Found dataset archive: {archive_path} ({os.path.getsize(archive_path) / (1024**2):.1f} MB)")
        print(f"⏳ Extracting to {OUTPUT_DIR}...")
        t0 = time.time()
        with zipfile.ZipFile(archive_path, 'r') as zf:
            zf.extractall('/content')
        print(f"✅ Extraction completed in {time.time() - t0:.1f}s!")
    else:
        print("⚠️ Dataset.zip not found automatically in /content.")
        print("👉 Please upload 'Dataset.zip' using the Colab Files panel (left sidebar),")
        print("   or mount Google Drive if you saved Dataset.zip in your Drive.")

train_dir = os.path.join(DATASET_DIR, "train")
val_dir   = os.path.join(DATASET_DIR, "val")
test_dir  = os.path.join(DATASET_DIR, "test")

if os.path.exists(train_dir):
    train_classes = sorted([d for d in os.listdir(train_dir) if os.path.isdir(os.path.join(train_dir, d))])
    print(f"📊 Dataset verified successfully: {len(train_classes)} unified classes found.")
    print(f"   Train classes preview: {train_classes[:5]} ...")
else:
    print("ℹ️ Note: Waiting for Dataset to be extracted into /content/Dataset before training.")
```

<div class='agro-card'>
  <div class='agro-badge'>⚙️ Colab Hyperparameters</div>
  <div class='agro-title'>Module 3: Google Colab GPU Hyperparameters</div>
  <div class='agro-subtitle'>Optimized for Tesla T4 / A100 GPU throughput with batch size 64-128, AMP mixed-precision, and AdamW optimizer with Cosine Annealing.</div>
</div>

```python
# ============================================================
# MODULE 3: HYPERPARAMETER CONFIGURATION (COLAB GPU)
# ============================================================

# Adjust batch size according to available VRAM
batch_size = 128 if (torch.cuda.is_available() and vram_gb >= 15.0) else 64

CONFIG = {
    'architecture': 'ConvNeXt-Tiny',
    'backbone': 'convnext_tiny.fb_in22k_ft_in1k',
    'img_size': 256,
    'batch_size': batch_size,
    'num_workers': 2,            # 2 workers matches Colab 2-vCPU allocation
    'pin_memory': torch.cuda.is_available(),
    'persistent_workers': True,
    'use_amp': torch.cuda.is_available(),
    'lr': 2e-4,
    'min_lr': 1e-6,
    'weight_decay': 1e-2,
    'epochs': 10,
    'label_smoothing': 0.1,
    'output_model': '/content/best_agroguard_model.pth'
}

# Agronomic Normalization Values (Computed over 51,320 agro leaf samples)
AGRO_MEAN = [0.4678, 0.4931, 0.3542]
AGRO_STD  = [0.2114, 0.1983, 0.2248]

print("⚙️ Google Colab Training Configuration:")
for k, v in CONFIG.items():
    print(f"   • {k:<18}: {v}")
```

<div class='agro-card'>
  <div class='agro-badge'>🌿 Field Robustness</div>
  <div class='agro-title'>Module 4: Sunlight, Glare, Shadow Augmentations & Class Balancing</div>
  <div class='agro-subtitle'>Engineered for robust outdoor deployment under variable daylight, waxy leaf reflections, and canopy shadows using WeightedRandomSampler.</div>
</div>

```python
# ============================================================
# MODULE 4: FIELD-ROBUST AUGMENTATION & BALANCED DATALOADERS
# ============================================================
from torchvision import datasets, transforms
from torch.utils.data import DataLoader, WeightedRandomSampler
from collections import Counter
import numpy as np

# Sunlight, glare, shade, and contrast transforms
train_transform = transforms.Compose([
    transforms.RandomResizedCrop(CONFIG['img_size'], scale=(0.70, 1.0), interpolation=transforms.InterpolationMode.BILINEAR),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.5),
    transforms.RandomAffine(degrees=20, translate=(0.08, 0.08), scale=(0.90, 1.10), shear=8),
    transforms.ColorJitter(brightness=0.35, contrast=0.35, saturation=0.30, hue=0.08),
    transforms.RandomAutocontrast(p=0.4),
    transforms.RandomEqualize(p=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=AGRO_MEAN, std=AGRO_STD)
])

eval_transform = transforms.Compose([
    transforms.Resize((CONFIG['img_size'], CONFIG['img_size'])),
    transforms.ToTensor(),
    transforms.Normalize(mean=AGRO_MEAN, std=AGRO_STD)
])

if os.path.exists(train_dir):
    train_dataset = datasets.ImageFolder(train_dir, transform=train_transform)
    val_dataset   = datasets.ImageFolder(val_dir, transform=eval_transform)
    test_dataset  = datasets.ImageFolder(test_dir, transform=eval_transform)

    classes = train_dataset.classes
    num_classes = len(classes)

    # Compute inverse class frequencies for WeightedRandomSampler
    targets = train_dataset.targets
    class_counts = Counter(targets)
    total_samples = len(targets)

    # Smoothed inverse weighting
    class_weights = {cls: (total_samples / (count ** 0.5)) for cls, count in class_counts.items()}
    sample_weights = np.array([class_weights[t] for t in targets], dtype=np.float32)
    sampler = WeightedRandomSampler(weights=sample_weights, num_samples=len(sample_weights), replacement=True)

    train_loader = DataLoader(
        train_dataset,
        batch_size=CONFIG['batch_size'],
        sampler=sampler,
        num_workers=CONFIG['num_workers'],
        pin_memory=CONFIG['pin_memory'],
        persistent_workers=CONFIG['persistent_workers']
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=CONFIG['batch_size'],
        shuffle=False,
        num_workers=CONFIG['num_workers'],
        pin_memory=CONFIG['pin_memory'],
        persistent_workers=CONFIG['persistent_workers']
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=CONFIG['batch_size'],
        shuffle=False,
        num_workers=CONFIG['num_workers'],
        pin_memory=CONFIG['pin_memory']
    )

    print(f"✅ DataLoaders initialized:")
    print(f"   • Training Samples   : {len(train_dataset):,}")
    print(f"   • Validation Samples : {len(val_dataset):,}")
    print(f"   • Test Samples       : {len(test_dataset):,}")
    print(f"   • Batches per Epoch  : {len(train_loader)}")
```

<div class='agro-card'>
  <div class='agro-badge'>🧠 ConvNeXt Architecture</div>
  <div class='agro-title'>Module 5: ConvNeXt-Tiny Neural Network Instantiation</div>
  <div class='agro-subtitle'>Loads state-of-the-art ConvNeXt-Tiny backbone pre-trained on ImageNet-22k and fine-tuned for 48 unified agricultural classes.</div>
</div>

```python
# ============================================================
# MODULE 5: CONVNEXT-TINY MODEL INSTANTIATION
# ============================================================
import timm

print("⏳ Instantiating ConvNeXt-Tiny unified vision model...")
model = timm.create_model(
    CONFIG['backbone'],
    pretrained=True,
    num_classes=48,
    drop_rate=0.2,
    drop_path_rate=0.1
)

model = model.to(device)
num_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"✅ Model Architecture: ConvNeXt-Tiny")
print(f"   • Trainable Parameters: {num_params:,}")
print(f"   • Target Output Classes: 48 unified classes")
print(f"   • Target Device        : {device}")
```

<div class='agro-card'>
  <div class='agro-badge'>⚡ Cloud Training</div>
  <div class='agro-title'>Module 6: Mixed-Precision (AMP) Cloud Training Loop</div>
  <div class='agro-subtitle'>Executes accelerated PyTorch training with automatic mixed precision, gradient scaling, AdamW optimizer, and Cosine Annealing learning rate.</div>
</div>

```python
# ============================================================
# MODULE 6: TRAINING & VALIDATION PIPELINE (AMP ENABLED)
# ============================================================
import torch.optim as optim
from torch.cuda.amp import autocast, GradScaler

criterion = nn.CrossEntropyLoss(label_smoothing=CONFIG['label_smoothing'])
optimizer = optim.AdamW(model.parameters(), lr=CONFIG['lr'], weight_decay=CONFIG['weight_decay'])
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=CONFIG['epochs'], eta_min=CONFIG['min_lr'])
scaler = GradScaler(enabled=CONFIG['use_amp'])

best_val_acc = 0.0
save_path = CONFIG['output_model']

print("=" * 80)
print(f"  AgroGuard v1.0 ConvNeXt-Tiny Training Commenced ({CONFIG['epochs']} Epochs)")
print("=" * 80)

for epoch in range(1, CONFIG['epochs'] + 1):
    t_start = time.time()
    model.train()
    train_loss, train_correct, train_total = 0.0, 0, 0
    
    for imgs, targets in train_loader:
        imgs, targets = imgs.to(device), targets.to(device)
        optimizer.zero_grad()
        
        with autocast(enabled=CONFIG['use_amp']):
            outputs = model(imgs)
            loss = criterion(outputs, targets)
            
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        
        train_loss += loss.item() * imgs.size(0)
        _, preds = outputs.max(1)
        train_correct += preds.eq(targets).sum().item()
        train_total += targets.size(0)
        
    scheduler.step()
    train_acc = 100.0 * train_correct / train_total
    avg_train_loss = train_loss / train_total
    
    # Validation Loop
    model.eval()
    val_loss, val_correct, val_total = 0.0, 0, 0
    with torch.no_grad():
        for imgs, targets in val_loader:
            imgs, targets = imgs.to(device), targets.to(device)
            with autocast(enabled=CONFIG['use_amp']):
                outputs = model(imgs)
                loss = criterion(outputs, targets)
                
            val_loss += loss.item() * imgs.size(0)
            _, preds = outputs.max(1)
            val_correct += preds.eq(targets).sum().item()
            val_total += targets.size(0)
            
    val_acc = 100.0 * val_correct / val_total
    avg_val_loss = val_loss / val_total
    epoch_time = time.time() - t_start
    
    print(f"Epoch [{epoch:02d}/{CONFIG['epochs']:02d}] ({epoch_time:.1f}s) | "
          f"Train Loss: {avg_train_loss:.4f} Acc: {train_acc:.2f}% | "
          f"Val Loss: {avg_val_loss:.4f} Acc: {val_acc:.2f}%")
          
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), save_path)
        print(f"   🏆 New Best Model Saved! Val Acc: {best_val_acc:.2f}% -> {save_path}")

print("=" * 80)
print(f"✅ Training Complete! Highest Validation Accuracy: {best_val_acc:.2f}%")
print("=" * 80)
```

<div class='agro-card'>
  <div class='agro-badge'>🧪 Test Metrics</div>
  <div class='agro-title'>Module 7: 3-View TTA Evaluation on Unseen Test Dataset</div>
  <div class='agro-subtitle'>Evaluates the trained model on 5,133 unseen test leaf images with Test-Time Augmentation (TTA), reporting Top-1 and Top-5 accuracy.</div>
</div>

```python
# ============================================================
# MODULE 7: 3-VIEW TTA TEST SET EVALUATION
# ============================================================
import numpy as np

def predict_tta(model, imgs):
    with torch.no_grad():
        with torch.cuda.amp.autocast(enabled=CONFIG['use_amp']):
            p1 = torch.softmax(model(imgs), dim=1)
            p2 = torch.softmax(model(torch.flip(imgs, dims=[3])), dim=1)
            p3 = torch.softmax(model(torch.flip(imgs, dims=[2])), dim=1)
        return (p1 + p2 + p3) / 3.0

if os.path.exists(save_path):
    model.load_state_dict(torch.load(save_path, map_location=device))
    model.eval()

    all_preds, all_targets = [], []
    top5_correct, total = 0, 0

    print("🔬 Running 3-View TTA on Test Dataset (5,133 unseen leaf images)...")
    with torch.no_grad():
        for imgs, targets in test_loader:
            imgs = imgs.to(device)
            probs = predict_tta(model, imgs).cpu()
            _, top1 = probs.max(1)
            _, top5 = probs.topk(5, dim=1)
            
            all_preds.extend(top1.numpy())
            all_targets.extend(targets.numpy())
            for t, k5 in zip(targets, top5):
                if t in k5:
                    top5_correct += 1
            total += len(targets)

    top1 = 100.0 * np.sum(np.array(all_preds) == np.array(all_targets)) / total
    top5 = 100.0 * top5_correct / total

    print("=" * 80)
    print(f"🏆 UNSEEN TEST EVALUATION RESULTS:")
    print(f"   • Unseen Test Samples : {total}")
    print(f"   • Top-1 Accuracy      : {top1:.2f}%")
    print(f"   • Top-5 Accuracy      : {top5:.2f}%")
    print("=" * 80)
```
